```python
from pathlib import Path
import pandas as pd
from IPython.display import display

task_name = "Task_2_Summary_Statistics"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "age"
values = df[column]
statistics = pd.Series({
    "Mean": values.mean(),
    "Median": values.median(),
    "Mode": values.mode().iloc[0],
    "Minimum": values.min(),
    "Maximum": values.max(),
    "Standard deviation": values.std(ddof=0),
}, name=column)
display(statistics.to_frame("Value"))
```
