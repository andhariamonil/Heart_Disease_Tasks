```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

task_name = "Task_12_Manual_Box_Plots"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "age"
values = df[column].dropna().to_numpy()
q1, median, q3 = np.percentile(values, [25, 50, 75])
iqr = q3 - q1
lower_fence, upper_fence = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = values[(values < lower_fence) | (values > upper_fence)]
whiskers = values[(values >= lower_fence) & (values <= upper_fence)]
box = {"med": median, "q1": q1, "q3": q3, "whislo": whiskers.min(), "whishi": whiskers.max(), "fliers": outliers}
print(f"Q1={q1:.2f}, median={median:.2f}, Q3={q3:.2f}, IQR={iqr:.2f}")
print(f"Outlier count: {len(outliers)}")
fig, ax = plt.subplots()
ax.bxp([box], showfliers=True)
ax.set_ylabel(column)
ax.set_title(f"Box plot of {column}")
plt.show()
```
