# Task 25 Custom CSV Storage

This notebook contains direct, self-contained code for the task.


```python
from pathlib import Path
import pandas as pd

active = Path.cwd().resolve()
output_dir = active if active.name.startswith('Task_25') else active.parent
possible = [active / 'heart.csv', active.parent / 'heart.csv']
csv_path = next((p for p in possible if p.exists()), Path('heart.csv'))
df = pd.read_csv(csv_path)
summary = df.groupby('target').agg(rows=('age', 'count'), avg_age=('age', 'mean')).reset_index()
out_path = output_dir / 'target_summary.csv'
summary.to_csv(out_path, index=False)
print('Saved file:', out_path)
print(summary.to_string(index=False))
```
