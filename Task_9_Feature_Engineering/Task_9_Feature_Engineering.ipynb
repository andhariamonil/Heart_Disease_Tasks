```python
from pathlib import Path
import pandas as pd
from IPython.display import display

task_name = "Task_9_Feature_Engineering"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

chol_percentile = df["chol"].rank(method="max", pct=True) * 100
thalach_percentile = df["thalach"].rank(method="max", pct=True) * 100
cholesterol_risk = df["chol"] / 200
heart_rate_risk = 220 / df["thalach"]
df["chol_percentile"] = chol_percentile
df["thalach_percentile"] = thalach_percentile
df["risk_factor"] = cholesterol_risk + heart_rate_risk + cholesterol_risk * heart_rate_risk + chol_percentile / 100 + (100 - thalach_percentile) / 100
display(df[["chol", "thalach", "chol_percentile", "thalach_percentile", "risk_factor"]].head())
```
