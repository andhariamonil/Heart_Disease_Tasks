```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

task_name = "Task_16_Chi_Square_Test"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "sex"
observed = pd.crosstab(df[column], df["target"])
expected = np.outer(observed.sum(axis=1), observed.sum(axis=0)) / observed.to_numpy().sum()
chi_square = (((observed.to_numpy() - expected) ** 2) / expected).sum()
degrees_of_freedom = (observed.shape[0] - 1) * (observed.shape[1] - 1)
critical_values_005 = {1: 3.841, 2: 5.991, 3: 7.815, 4: 9.488, 5: 11.070, 6: 12.592, 7: 14.067, 8: 15.507}
critical_value = critical_values_005.get(degrees_of_freedom)
print(f"Chi-square statistic: {chi_square:.3f}")
print(f"Degrees of freedom: {degrees_of_freedom}")
if critical_value is not None:
    print(f"Critical value at 0.05: {critical_value:.3f}")
    print("Association detected" if chi_square > critical_value else "No significant association detected")
display(observed)
```
