```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

task_name = "Task_3_Missing_Value_Imputation"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

print("Missing values before imputation:")
display(df.isna().sum().to_frame("Missing count"))

imputed_df = df.copy()
imputed_df.loc[:2, "chol"] = np.nan
imputed_df.loc[:2, "sex"] = np.nan
numerical_columns = ["age", "trestbps", "chol", "thalach", "oldpeak"]
categorical_columns = ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal", "target"]
for column in numerical_columns:
    if imputed_df[column].isna().any():
        imputed_df[column] = imputed_df[column].fillna(imputed_df[column].median())
for column in categorical_columns:
    if imputed_df[column].isna().any():
        imputed_df[column] = imputed_df[column].fillna(imputed_df[column].mode().iloc[0])
print("Demo rows after median/mode imputation:")
display(imputed_df.head())
```
