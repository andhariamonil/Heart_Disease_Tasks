```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

task_name = "Task_11_Manual_Histograms"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "age"
values = df[column].dropna().tolist()
number_of_bins = 10
minimum, maximum = min(values), max(values)
bin_width = (maximum - minimum) / number_of_bins
frequencies = [0] * number_of_bins
for value in values:
    bin_index = min(int((value - minimum) / bin_width), number_of_bins - 1)
    frequencies[bin_index] += 1
bin_starts = [minimum + index * bin_width for index in range(number_of_bins)]
plt.bar(bin_starts, frequencies, width=bin_width, align="edge", edgecolor="black")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.title(f"Manually binned histogram of {column}")
plt.show()
```
