```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

task_name = "Task_20_VIF_Analysis"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)
features = ["age", "trestbps", "chol", "thalach", "oldpeak", "ca"]
results = []
for feature in features:
    target = df[feature].to_numpy(dtype=float)
    predictors = df[[name for name in features if name != feature]].to_numpy(dtype=float)
    design = np.column_stack([np.ones(len(predictors)), predictors])
    coefficients = np.linalg.lstsq(design, target, rcond=None)[0]
    predictions = design @ coefficients
    total = ((target - target.mean()) ** 2).sum()
    r_squared = 1 - ((target - predictions) ** 2).sum() / total if total else 0
    vif = float("inf") if r_squared >= 1 else 1 / (1 - r_squared)
    results.append({"Feature": feature, "R-squared": r_squared, "VIF": vif})
display(pd.DataFrame(results).round(4))
```
