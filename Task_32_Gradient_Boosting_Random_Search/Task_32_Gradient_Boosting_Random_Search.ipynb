# Task 32 Gradient Boosting Random Search

This notebook contains direct, self-contained code for the task.


```python
from pathlib import Path
import sys
import random
import numpy as np

active = Path.cwd().resolve()
tasks_root = active if (active / 'manual_ml.py').exists() else active.parent
if str(tasks_root) not in sys.path:
    sys.path.insert(0, str(tasks_root))

from manual_ml import GradientBoostingScratch, load_data, stratified_split

possible = [active / 'heart.csv', tasks_root / 'heart.csv', active.parent / 'heart.csv']
csv_path = next((p for p in possible if p.exists()), Path('heart.csv'))
X, y, features, df = load_data(csv_path)
X_train, X_test, y_train, y_test = stratified_split(X, y, test_size=0.25, seed=18)

rng = random.Random(9)
results = []
for _ in range(6):
    lr = round(rng.uniform(0.05, 0.35), 3)
    n_estimators = rng.randint(10, 40)
    model = GradientBoostingScratch(n_estimators=n_estimators, learning_rate=lr)
    model.fit(X_train, y_train)
    acc = float((model.predict(X_test) == y_test).mean())
    results.append((acc, lr, n_estimators))

best = max(results, key=lambda item: item[0])
print('Best random-search result:', best)
print('Top 3 results:', sorted(results, reverse=True)[:3])
```
