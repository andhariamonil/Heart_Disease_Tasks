```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

task_name = "Task_5_Data_Visualization"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)
numerical_columns = ["age", "trestbps", "chol", "thalach", "oldpeak", "ca"]

fig, axes = plt.subplots(2, 3, figsize=(12, 7))
for axis, column in zip(axes.flat, numerical_columns):
    axis.hist(df[column].dropna(), bins=10, edgecolor="black")
    axis.set_title(column)
    axis.set_ylabel("Frequency")
fig.tight_layout()
plt.show()

display(df[numerical_columns].describe())
```
