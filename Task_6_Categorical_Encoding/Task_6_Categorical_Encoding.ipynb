```python
from pathlib import Path
import pandas as pd
from IPython.display import display

task_name = "Task_6_Categorical_Encoding"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "cp"
categories = list(pd.unique(df[column]))
category_codes = {category: code for code, category in enumerate(categories)}
encoded_df = df[[column]].copy()
encoded_df[f"{column}_encoded"] = encoded_df[column].map(category_codes)
print("Category to integer mapping:")
display(pd.DataFrame(category_codes.items(), columns=["Category", "Code"]))
display(encoded_df.head())
```
