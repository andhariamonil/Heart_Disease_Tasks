```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

task_name = "Task_14_Missing_Value_Heatmap"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)
missing_grid = df.isna().astype(int)
missing_count = int(missing_grid.to_numpy().sum())
print(f"Missing values: {missing_count}")
if missing_count == 0:
    print("No missing values were found in the original dataset.")
fig, ax = plt.subplots(figsize=(10, 5))
image = ax.imshow(missing_grid, aspect="auto", interpolation="nearest", cmap="Reds")
ax.set_xticks(range(len(df.columns)), df.columns, rotation=90)
ax.set_xlabel("Columns")
ax.set_ylabel("Rows")
ax.set_title("Missing-value heatmap")
fig.colorbar(image, ax=ax, label="Missing (1) / Present (0)")
fig.tight_layout()
plt.show()
```
