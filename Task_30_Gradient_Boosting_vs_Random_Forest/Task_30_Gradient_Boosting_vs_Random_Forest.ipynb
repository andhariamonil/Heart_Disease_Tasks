# Task 30 Gradient Boosting vs Random Forest

This notebook contains direct, self-contained code for the task.


```python
from pathlib import Path
import sys
import numpy as np

active = Path.cwd().resolve()
tasks_root = active if (active / 'manual_ml.py').exists() else active.parent
if str(tasks_root) not in sys.path:
    sys.path.insert(0, str(tasks_root))

from manual_ml import GradientBoostingScratch, RandomForestScratch, load_data, stratified_split

possible = [active / 'heart.csv', tasks_root / 'heart.csv', active.parent / 'heart.csv']
csv_path = next((p for p in possible if p.exists()), Path('heart.csv'))
X, y, features, df = load_data(csv_path)
X_train, X_test, y_train, y_test = stratified_split(X, y, test_size=0.25, seed=11)

rf = RandomForestScratch(n_trees=10, max_depth=4)
rf.fit(X_train, y_train, features)
rf_acc = float((rf.predict(X_test) == y_test).mean())

gb = GradientBoostingScratch(n_estimators=20, learning_rate=0.15)
gb.fit(X_train, y_train)
gb_acc = float((gb.predict(X_test) == y_test).mean())

print('Random Forest accuracy:', round(rf_acc, 4))
print('Gradient Boosting accuracy:', round(gb_acc, 4))
```
