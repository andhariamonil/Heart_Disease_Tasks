```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

task_name = "Task_19_Outlier_Transformation"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "age"
values = df[column].astype(float)
q1, q3 = values.quantile([0.25, 0.75])
winsorized = values.clip(lower=q1, upper=q3)
log_transformed = np.log(values.where(values > 0, 1))
comparison = pd.DataFrame({"Original": values, "Winsorized": winsorized, "Log transformed": log_transformed})
display(comparison.head())
display(comparison.agg(["mean", "std"]).T.rename(columns={"mean": "Mean", "std": "Standard deviation"}))
```
