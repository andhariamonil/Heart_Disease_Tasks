```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

task_name = "Task_21_Decision_Tree_Feature_Importance"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)
features = ["age", "trestbps", "chol", "thalach", "oldpeak", "ca"]
target = df["target"].to_numpy()

def gini(labels):
    _, counts = np.unique(labels, return_counts=True)
    probabilities = counts / len(labels)
    return 1 - np.sum(probabilities ** 2)

parent_gini = gini(target)
results = []
for feature in features:
    values = np.sort(df[feature].unique())
    thresholds = (values[:-1] + values[1:]) / 2
    best_gain, best_threshold = 0.0, None
    for threshold in thresholds:
        left = target[df[feature].to_numpy() <= threshold]
        right = target[df[feature].to_numpy() > threshold]
        weighted_gini = len(left) / len(target) * gini(left) + len(right) / len(target) * gini(right)
        gain = parent_gini - weighted_gini
        if gain > best_gain:
            best_gain, best_threshold = gain, threshold
    results.append({"Feature": feature, "Best threshold": best_threshold, "Gini gain": best_gain})
display(pd.DataFrame(results).sort_values("Gini gain", ascending=False).round(6))
```
