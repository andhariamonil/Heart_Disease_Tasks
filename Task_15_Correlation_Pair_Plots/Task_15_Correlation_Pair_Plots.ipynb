```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

task_name = "Task_15_Correlation_Pair_Plots"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)
numerical_columns = ["age", "trestbps", "chol", "thalach", "oldpeak", "ca"]
correlations = df[numerical_columns].corr()
threshold = 0.5
pairs = [(first, second, correlations.loc[first, second]) for index, first in enumerate(numerical_columns) for second in numerical_columns[index + 1:] if abs(correlations.loc[first, second]) >= threshold]
print(f"Pairs with |correlation| >= {threshold}: {len(pairs)}")
display(correlations)
for first, second, correlation in pairs:
    ax = df.plot.scatter(x=first, y=second, alpha=0.5)
    ax.set_title(f"{first} vs {second} (r={correlation:.3f})")
    plt.show()
```
