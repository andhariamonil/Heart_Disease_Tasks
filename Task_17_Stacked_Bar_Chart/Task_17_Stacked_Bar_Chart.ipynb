```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

task_name = "Task_17_Stacked_Bar_Chart"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)
proportions = pd.crosstab(df["sex"], df["target"], normalize="index").rename(columns={0: "No disease", 1: "Heart disease"})
ax = proportions.plot(kind="bar", stacked=True, figsize=(7, 5), color=["#79a9dc", "#e87961"])
ax.set_xlabel("Sex category")
ax.set_ylabel("Proportion")
ax.set_title("Heart-disease proportion by sex category")
ax.legend(title="Target")
plt.tight_layout()
plt.show()
display(proportions)
```
