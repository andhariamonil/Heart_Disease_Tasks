# Task 27 Decision Tree Grid Search

This notebook contains direct, self-contained code for the task.


```python
from pathlib import Path
import sys
import numpy as np

active = Path.cwd().resolve()
tasks_root = active if (active / 'manual_ml.py').exists() else active.parent
if str(tasks_root) not in sys.path:
    sys.path.insert(0, str(tasks_root))

from manual_ml import DecisionTreeScratch, load_data, stratified_split

possible = [active / 'heart.csv', tasks_root / 'heart.csv', active.parent / 'heart.csv']
csv_path = next((p for p in possible if p.exists()), Path('heart.csv'))
X, y, features, df = load_data(csv_path)
X_train, X_test, y_train, y_test = stratified_split(X, y, test_size=0.2, seed=5)

results = []
for depth in [2, 3, 4, 5]:
    for min_split in [4, 6, 8]:
        model = DecisionTreeScratch(max_depth=depth, min_samples_split=min_split)
        model.fit(X_train, y_train, features)
        preds = model.predict(X_test)
        acc = float((preds == y_test).mean())
        results.append((acc, depth, min_split))

best = max(results, key=lambda item: item[0])
print('Best result:', best)
print('All scores:', sorted(results, reverse=True)[:5])
```
