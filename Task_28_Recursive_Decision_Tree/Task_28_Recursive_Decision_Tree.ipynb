# Task 28 Recursive Decision Tree

This notebook contains direct, self-contained code for the task.


```python
from pathlib import Path
import sys
import numpy as np

active = Path.cwd().resolve()
tasks_root = active if (active / 'manual_ml.py').exists() else active.parent
if str(tasks_root) not in sys.path:
    sys.path.insert(0, str(tasks_root))

from manual_ml import best_tree_split, load_data

possible = [active / 'heart.csv', tasks_root / 'heart.csv', active.parent / 'heart.csv']
csv_path = next((p for p in possible if p.exists()), Path('heart.csv'))
X, y, features, df = load_data(csv_path)

# Build a tiny recursive tree structure from a selected subset of features
subset = X[:, :4]
feature, threshold, gain = best_tree_split(subset, y, features=np.arange(subset.shape[1]), criterion='gini', min_samples_split=10)
print('Root feature index:', feature)
print('Root threshold:', round(float(threshold), 3))
print('Information gain:', round(float(gain), 6))
print('Feature names used:', features[:4])
```
