```python
from pathlib import Path
import pandas as pd
from IPython.display import display

task_name = "Task_7_Normalization_Standardization"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "age"
values = df[column].astype(float)
median = values.median()
mad = (values - median).abs().median()
scaled_df = pd.DataFrame({
    "Original": values,
    "Min-max": (values - values.min()) / (values.max() - values.min()),
    "Z-score": (values - values.mean()) / values.std(ddof=0),
    "Robust": (values - median) / mad,
})
print(f"Column: {column}; skewness: {values.skew():.3f}")
display(scaled_df.head(10))
```
