```python
from pathlib import Path
import pandas as pd
from IPython.display import display

task_name = "Task_8_One_Hot_Encoding"
candidates = [Path("heart.csv"), Path(task_name) / "heart.csv", Path("Heart_Disease_Tasks") / task_name / "heart.csv"]
csv_path = next(path for path in candidates if path.is_file())
df = pd.read_csv(csv_path)

column = "cp"
encoded_df = pd.get_dummies(df, columns=[column], prefix=column, dtype=int)
print(f"One-hot encoded column: {column}")
display(encoded_df.head())
```
