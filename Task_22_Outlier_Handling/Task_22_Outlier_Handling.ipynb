# Task 22 Outlier Handling

This notebook contains direct, self-contained code for the task.


```python
from pathlib import Path
import numpy as np
import pandas as pd

active = Path.cwd().resolve()
tasks_root = active if (active / 'manual_ml.py').exists() else active.parent
if str(tasks_root) not in __import__('sys').path:
    __import__('sys').path.insert(0, str(tasks_root))

possible = [active / 'heart.csv', tasks_root / 'heart.csv', active.parent / 'heart.csv']
csv_path = next((p for p in possible if p.exists()), Path('heart.csv'))
df = pd.read_csv(csv_path)
print('Rows:', len(df))
print('Columns:', list(df.columns[:5]), '...')

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
selected = numeric_cols[:4]
Q1 = df[selected].quantile(0.25)
Q3 = df[selected].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outlier_counts = ((df[selected] < lower) | (df[selected] > upper)).sum()
print(outlier_counts.to_dict())

cleaned = df.copy()
for col in selected:
    mask = (cleaned[col] < lower[col]) | (cleaned[col] > upper[col])
    cleaned.loc[mask, col] = cleaned[col].clip(lower=lower[col], upper=upper[col])
print(cleaned.head(3).to_string(index=False))
```
